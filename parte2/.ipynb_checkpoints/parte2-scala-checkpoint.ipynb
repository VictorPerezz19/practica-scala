## Ejercicio 1 — Variables, tipos e inferencia
**Explicación:** 
A continuación, crearemos las variables de un jugador de cartas. En la Versión A indicaremos los tipos de datos a mano (String, Int, Double, Boolean). En la Versión B omitiremos los tipos para que el compilador de Scala los deduzca automáticamente por su cuenta.

````scala
// Versión A: Tipos explícitos
val nombreA: String = "Víctor"
var partidasA: Int = 12
var puntuacionA: Int = 350
var porcentajeVictoriasA: Double = 75.5
var activoA: Boolean = true

// Versión B: Tipos inferidos
val nombreB = "Víctor"
var partidasB = 12
var puntuacionB = 350
var porcentajeVictoriasB = 75.5
var activoB = true

// Mostrar los valores por pantalla
println(s"Versión A -> Jugador: $nombreA, Partidas: $partidasA, Puntuación: $puntuacionA, Victorias: $porcentajeVictoriasA%, Activo: $activoA")
println(s"Versión B -> Jugador: $nombreB, Partidas: $partidasB, Puntuación: $puntuacionB, Victorias: $porcentajeVictoriasB%, Activo: $activoB")
````

**Documentación y Conclusiones:**

* **Tipos inferidos:** Scala ha detectado automáticamente que el texto es un `String`, los números enteros son `Int`, el número con decimales es un `Double` y el valor de verdadero/falso es un `Boolean`.
* **Diferencia entre versiones:** La Versión B es mucho más limpia, directa y rápida de escribir. Al quitar el `: Tipo`, el código queda menos saturado visualmente y delega ese trabajo de comprobación al compilador de Scala.
* **Uso de val vs var:** 
  * `nombre` debe ser **`val`** (inmutable), ya que la identidad del  jugador no va a cambiar durante la partida.
  * `partidas`, `puntuacion`, `porcentajeVictorias` y `activo` deben ser **`var`** (mutables), puesto que son valores que necesariamente van a subir, bajar o cambiar de estado a medida que el jugador avance en el juego.

````scala
val valorDouble: Double = 3.14159265358979323846264338327
val valorFloat: Float = 3.14159265358979323846264338327f
println(valorDouble)
println(valorFloat)

val numeroEstudiantes: Int = 25
val haAprobado: Boolean = true
val asignatura: String = "Scala"
````

**Documentación y Conclusiones (Ejercicio 3):**

## Ejercicio 4 — Función para determinar si una mano se pasa de 21
**Explicación:** Vamos a crear una función pura llamada `bust` que evaluará si la puntuación recibida supera los 21 puntos. Comprobaremos su funcionamiento con los valores de prueba requeridos (18, 21, 22 y 30).

````scala
def bust(puntuacion: Int): Boolean = 
  if (puntuacion > 21) true else false

// Pruebas uniendo el texto y la función con el símbolo +
println("¿Con 18 se pasa? " + bust(18))
println("¿Con 21 se pasa? " + bust(21))
println("¿Con 22 se pasa? " + bust(22))
println("¿Con 30 se pasa? " + bust(30))
````

**Documentación y Conclusiones (Ejercicio 4):**

* **Cumplimiento del requisito:** La función `bust` no modifica variables externas. Únicamente toma el valor de entrada (`puntuacion`), lo evalúa y devuelve el resultado, cumpliendo el principio de evitar efectos secundarios.
* **Simplificación del código:** En lugar de utilizar una estructura condicional (`if (puntuacion > 21) true else false`), Scala permite devolver directamente la expresión lógica `puntuacion > 21`. Como esta operación matemática ya genera por sí misma un valor `Boolean`, el código queda mucho más limpio, directo y profesional.

## Ejercicio 5 — Comparación de dos manos

Define una función:

```scala
maxHand
```

que reciba dos valores enteros y devuelva el mayor de los dos utilizando `if` y `else`.

Prueba la función al menos con:

```
17 y 19
20 y 18
21 y 21
```

### Después

Explica qué sucede cuando ambas manos tienen el mismo valor.

Modifica la función si consideras necesario mejorar el comportamiento para ese caso y documenta tu decisión.

````scala
def maxHand(handA: Int, handB: Int): Int = {
  if (handA > handB) {
    handA
  } else {
    handB
  }
}

println("Entre 17 y 19, la mayor es: " + maxHand(17, 19))
println("Entre 20 y 18, la mayor es: " + maxHand(20, 18))
println("Entre 21 y 21, la mayor es: " + maxHand(21, 21))
````

**Documentación y Conclusiones (Ejercicio 5):**

* **¿Qué sucede cuando ambas manos tienen el mismo valor (ej. 21 y 21)?**
  Cuando ambas manos valen lo mismo, la condición `if (handA > handB)` evalúa `21 > 21`, lo cual es `false`. Por lo tanto, el flujo salta directamente al bloque `else` y devuelve `handB`. 

* **Decisión de diseño (¿Es necesario modificarla?):**
  No considero necesario modificar la función añadiendo un caso extra para los empates (como un `else if (handA == handB)`). Dado que ambos números son idénticos, devolver cualquiera de los dos sigue siendo la respuesta correcta a la pregunta "¿cuál es el valor máximo?". Al dejarla así, la función hace exactamente lo que debe, gastando menos memoria y con menos líneas de código.

## Ejercicio 6 — Decidir el ganador de una partida
**Explicación:** Vamos a crear la función `ganador`, la cual reutiliza las funciones anteriores (`bust` y `maxHand`). Mediante una estructura condicional `if / else if / else` aplicaremos las reglas del juego para determinar qué puntuación es la ganadora, devolviendo 0 si ambos se pasan de 21.

````scala
def ganador(handA: Int, handB: Int): Int = {
    if (bust(handA) && bust(handB)) {
    0
  } else if (bust(handA)) {
    handB
  } else if (bust(handB)) {
    handA
  } else {
    maxHand(handA, handB)
  }
}

println(ganador(26, 20))
println(ganador(18, 22))
println(ganador(24, 25))
println(ganador(17, 19))
println(ganador(21, 20))
````

**Documentación y Conclusiones (Condiciones cumplidas por caso):**

1. **Prueba 26 y 20:** Solo `handA` se pasa. Falla el primer `if` y se cumple la condición `else if (bust(handA))`. Devuelve la puntuación de `handB` (20).
2. **Prueba 18 y 22:** Solo `handB` se pasa. Falla el primer y segundo condicional, y se cumple la condición `else if (bust(handB))`. Devuelve la puntuación de `handA` (18).
3. **Prueba 24 y 25:** Ambas superan los 21 puntos. Se cumple la primera condición de todas: `if (bust(handA) && bust(handB))`. Devuelve 0.
4. **Prueba 17 y 19:** Ninguna se pasa. No se cumple ningún `if` ni `else if`, por lo que el flujo cae en el bloque `else` final. Llama a la función `maxHand` y devuelve 19.
5. **Prueba 21 y 20:** Ninguna se pasa. Al igual que el caso anterior, cae en el bloque `else` final, evalúa quién tiene la puntuación más alta y devuelve 21.

## Ejercicio 7 — Arrays y mutabilidad
**Explicación:** En este ejercicio crearemos un array de nombres de jugadores. Observaremos cómo, aunque el array se declare como una constante (`val`), podemos modificar su contenido interno. Finalmente, intentaremos forzar un error de tipos introduciendo un número en un array de textos.

````scala
val jugadores = Array("Alex", "Chen", "Marta")

// Mostrar el array original
println(jugadores.mkString(", "))

// Sustituir "Alex" (posición 0) por "Sindhu"
jugadores(0) = "Sindhu"

// Mostrar nuevamente el array modificado
println(jugadores.mkString(", "))

// Intentar asignar el número 500 a una posición (esto generará un error de compilación)
jugadores(1) = 500
````

**Documentación y Conclusiones (Ejercicio 7):**

* **Por qué los elementos pueden cambiar aunque se use `val`:** 
  La palabra reservada `val` hace que la *referencia* a la variable sea inmutable (es decir, la variable `jugadores` siempre apuntará al mismo Array en la memoria). Sin embargo, la estructura de datos `Array` en sí misma es mutable en Scala, lo que permite modificar los elementos que hay dentro de sus posiciones.
* **Por qué Scala no permite introducir el valor 500:** 
  Scala es un lenguaje de tipado estricto. Al crear el array con texto, el compilador infirió automáticamente que es un `Array[String]`. Por tanto, da un error de tipo (`type mismatch`) si intentamos meter un número entero (`Int`, como el 500) en una caja diseñada solo para texto.
* **Diferencia entre reasignar la variable y modificar un elemento:** 
  - *Reasignar la variable:* Sería intentar hacer `jugadores = Array("Pepe", "Luis")`. Esto está prohibido porque `jugadores` es `val` y no puede apuntar a un objeto nuevo.
  - *Modificar un elemento:* Es hacer `jugadores(0) = "Sindhu"`. Esto sí está permitido porque no cambiamos la identidad del array, solo actualizamos su contenido interno.

## Ejercicio 9 — Recorrer un Array con while
**Explicación:** Vamos a recorrer un array de puntuaciones paso a paso utilizando un bucle clásico de estilo imperativo (`while`). En cada vuelta del bucle consultaremos el valor de la mano y comprobaremos si devuelve `true` o `false` al aplicarle la función `bust`.

````scala
val manos = Array(17, 24, 21, 19, 26)
var i = 0

while (i < manos.length) {
  val valor = manos(i)
  val sePasa = bust(valor)
  println("Mano: " + valor + "  ¿Se pasa de 21?: " + sePasa)
  i += 1
}
````

**Documentación y Conclusiones (Ejercicio 9):**

* **Por qué el contador debe ser mutable (`var`):** 
  El contador `i` necesita ser de tipo `var` porque su valor cambia y se actualiza constantemente en cada iteración del bucle (`0, 1, 2, 3...`). Si fuera una constante inmutable (`val`), nos daría un error de compilación al intentar incrementarlo.
* **Qué ocurriría si no incrementaras `i`:** 
  Se provocaría un **bucle infinito**. El programa se quedaría atrapado evaluando permanentemente la primera posición del array (`manos(0)`), imprimiendo resultados sin parar hasta bloquear o saturar la ejecución del entorno.
* **Elementos del ejercicio que corresponden al estilo imperativo:** 
  Pertenecen al paradigma imperativo la utilización del bucle `while`, la declaración de una variable de estado externa para controlar el bucle (`var i = 0`), y la modificación manual y secuencial de dicho estado paso a paso (`i += 1`).

## Ejercicio 10 — Listas e inmutabilidad
**Explicación:** Vamos a trabajar con listas inmutables (`List`). Crearemos una lista inicial de jugadores, le añadiremos un elemento nuevo utilizando el operador de construcción `::` (cons), comprobaremos que la lista original permanece intacta y utilizaremos funciones predefinidas como `length` y `reverse`.

````scala
val jugadores = List("Alex", "Chen", "Marta")

println("Lista original: " + jugadores.mkString(", "))

val nuevosJugadores = "Sindhu" :: jugadores

println("Lista nueva (con Sindhu): " + nuevosJugadores.mkString(", "))
println("Lista original intacta: " + jugadores.mkString(", "))

println("Longitud de la nueva lista: " + nuevosJugadores.length)

println("Lista nueva invertida: " + nuevosJugadores.reverse.mkString(", "))
````

**Documentación y Conclusiones (Ejercicio 10):**

* **Por qué la operación con `::` produce una lista nueva en lugar de modificar la original:**
  En Scala, las listas (`List`) son estructuras de datos intrínsecamente **inmutables**. Esto significa que sus elementos no pueden ser alterados una vez creados. Cuando utilizamos el operador `::` (conocido como *cons*), Scala no modifica la lista preexistente en la memoria; en su lugar, crea un nuevo objeto de tipo lista que apunta al nuevo elemento añadido en la cabecera, reutilizando eficientemente el resto de la lista original. Por esta razón, la lista `jugadores` se mantiene exactamente igual mientras que `nuevosJugadores` almacena el resultado modificado.

  * **Uso de la función `.mkString(", ")`:** 
  Para evitar que al imprimir un Array o una List por pantalla se muestren referencias internas de memoria o códigos ilegibles, utilizamos `.mkString(...)`. El texto que se pasa entre paréntesis (en este caso, `", "`) actúa como **delimitador o separador** entre los elementos, transformando toda la colección en un único texto limpio, ordenado y fácil de leer.

## Ejercicio 11 — Construcción y concatenación de listas
**Explicación:** Construiremos dos listas de manera explícita utilizando `Nil` y el operador `::`. A continuación, las uniremos utilizando el operador de concatenación `:::` y comprobaremos mediante la salida por pantalla que ninguna de las dos listas originales sufre alteraciones.

````scala
val lista1 = "Ana" :: "Luis" :: "Marta" :: Nil

val lista2 = "Pedro" :: "Sofia" :: Nil

val listaConcatenada = lista1 ::: lista2

println("Primera lista: " + lista1.mkString(", "))
println("Segunda lista: " + lista2.mkString(", "))
println("Lista resultante (concatenada): " + listaConcatenada.mkString(", "))
````

**Documentación y Conclusiones (Ejercicio 11):**

* **Demostración de inmutabilidad en la concatenación:** 
  Al imprimir los resultados por pantalla, se puede comprobar claramente que la `lista1` ("Ana", "Luis", "Marta") y la `lista2` ("Pedro", "Sofia") se mantienen exactamente igual antes y después de la operación. El operador de concatenación `:::` crea un **nuevo objeto lista** que fusiona ambos contenidos, respetando la naturaleza inmutable de las listas en Scala y garantizando que los datos originales no se corrompan ni se modifiquen.

* **Uso de la función `.mkString(", ")`:** 
  Para evitar que al imprimir un Array o una List por pantalla se muestren referencias internas de memoria o códigos ilegibles, utilizamos `.mkString(...)`. El texto que se pasa entre paréntesis (en este caso, `", "`) actúa como **delimitador o separador** entre los elementos, transformando toda la colección en un único texto limpio, ordenado y fácil de leer.

## Ejercicio 12 — Operadores relacionales y lógicos
**Explicación:** Evaluaremos diferentes condiciones lógicas y comparativas utilizando las variables de puntuación `handA` (18), `handB` (21) y `handC` (25), aplicando los operadores relacionales y lógicos básicos de Scala.

````scala
val handA = 18
val handB = 21
val handC = 25

// ¿Es handA mayor que handB?
println(handA > handB)

// ¿Es handB igual a 21?
println(handB == 21)

// ¿Es handC diferente de 21?
println(handC != 21)

// ¿Son handA y handB menores o iguales a 21?
println(handA <= 21 && handB <= 21)

// ¿Alguna de las manos handA o handC supera 21?
println(handA > 21 || handC > 21)

// ¿No se ha pasado handB de 21?
println(!(handB > 21))
````

**Documentación y Conclusiones (Ejercicio 12):**

A continuación se detalla la tabla resumen con las expresiones evaluadas, su resultado booleano y la explicación lógica de cada una:

| Expresión | Resultado | Explicación |
| :--- | :---: | :--- |
| `handA > handB` | `false` | 18 no es mayor estricto que 21. |
| `handB == 21` | `true` | Comprueba si 21 es exactamente igual a 21. |
| `handC != 21` | `true` | Verifica que 25 sea diferente de 21, lo cual se cumple. |
| `handA <= 21 && handB <= 21` | `true` | Ambas condiciones son ciertas (18 y 21 son menores o iguales que 21). |
| `handA > 21 \|\| handC > 21` | `true` | Aunque `handA` no supera 21, `handC` (25) sí lo hace; al ser un "o" (`\|\|`), basta con que una se cumpla. |
| `!(handB > 21)` | `true` | Como `handB > 21` es falso (21 no es mayor que 21), el operador de negación `!` invierte el valor lógico a verdadero. |

## Ejercicio 13 — foreach y funciones como valores
**Explicación:** Vamos a recorrer un array de puntuaciones utilizando dos enfoques distintos para comprobar si cada mano se pasa de 21: la **Versión A** mediante un bucle imperativo (`while`) y la **Versión B** utilizando el método funcional `.foreach`.

````scala
val manos = Array(17, 24, 21, 26, 18)

//while 
println("--- Versión A (Bucle while) ---")
var i = 0
while (i < manos.length) {
  println(bust(manos(i)))
  i += 1
}

//foreach
println("--- Versión B (foreach funcional) ---")
manos.foreach(mano => println(bust(mano)))
````

**Documentación y Conclusiones (Comparativa Versión A vs Versión B):**

* **¿Cuál necesita un contador (`i`)?** 
  La **Versión A (while)** es la única que necesita un contador numérico explícito (`i`), ya que el programador debe gestionar manualmente el índice para acceder a cada posición del array.
* **¿Cuál necesita una variable mutable (`var`) para recorrer la colección?** 
  Nuevamente, la **Versión A (while)** requiere declarar una variable de tipo `var` (`var i = 0`) porque su valor tiene que ir incrementándose en cada vuelta (`i += 1`). Por el contrario, la Versión B opera con colecciones y funciones de orden superior sin necesidad de alterar ninguna variable externa.
* **¿Cuál se aproxima más al estilo funcional presentado en el curso?** 
  La **Versión B (foreach)** se aproxima de manera directa al paradigma funcional. En lugar de indicar *cómo* debe recorrerse la estructura paso a paso (control de índices, saltos y bucles), el enfoque funcional se centra en *qué* se quiere hacer, aplicando una función anónima directamente sobre cada elemento de la colección de forma declarativa, limpia y sin efectos secundarios de mutabilidad de estado.

## Ejercicio 14 — Efectos secundarios y estilo de programación
**Explicación:** Vamos a analizar el concepto de **efecto secundario** (side effect) comparando una función impura que modifica una variable externa (`sumarAlTotal`) frente a una función pura que opera únicamente con sus parámetros de entrada y devuelve un resultado sin alterar el exterior (`sumar`).

````scala
 
var total = 0

def sumarAlTotal(valor: Int): Unit = {
  total = total + valor
}

sumarAlTotal(5)
println(total)

sumarAlTotal(10)
println(total)

sumarAlTotal(3)
println(total)

// Versión 2
def sumar(a: Int, b: Int): Int = {
  a + b
}

println(sumar(5, 10))
println(sumar(20, 5))
````

## Ejercicio 14 — Efectos secundarios y estilo de programación
**Explicación:** Analizamos el concepto de **efecto secundario** comparando una función impura que modifica una variable externa (`sumarAlTotal`) frente a una función pura que opera solo con sus parámetros de entrada y devuelve un resultado sin alterar el exterior (`sumar`).

**Documentación y Conclusiones:**

* **Por qué `sumarAlTotal` modifica una variable externa:** Modifica la variable `total` porque está declarada fuera de su ámbito local. Al ser de tipo `var`, la función accede a ella y reescribe su valor.
* **Identificación del efecto secundario:** El efecto secundario consiste en que la función altera de forma oculta el estado del programa (`total`), haciendo que su comportamiento dependa del historial de llamadas y no solo de los parámetros de entrada.

| Característica | `sumarAlTotal` (Versión 1) | `sumar` (Versión 2) |
| :--- | :---: | :---: |
| **Modifica datos externos** | Sí | No |
| **Devuelve un resultado calculado** | No (devuelve `Unit`) | Sí (devuelve el entero resultante) |
| **Utiliza efecto secundario** | Sí | No |
| **Estilo predominante** | Imperativo / Impuro | Funcional / Puro |

## Ejercicio 15 — Programa integrado: torneo de Twenty-One
**Explicación:** Desarrollaremos un programa completo que combina colecciones inmutables (`List`), arrays mutables (`Array`), estructuras de control (`while`, `if/else`), funciones personalizadas (`bust` y `mejorPuntuacionValida`) y análisis de múltiples rondas para un torneo de cuatro jugadores.

````scala
val jugadores = List("Alex", "Chen", "Marta", "Sindhu")
val ronda1 = Array(18, 24, 21, 20)
val ronda2 = Array(22, 19, 20, 21)

// buscar la mejor puntuación válida (<= 21)
def mejorPuntuacion(manos: Array[Int]): Int = {
  var mejor = 0
  var i = 0
  while (i < manos.length) {
    val puntos = manos(i)
    if (puntos <= 21 && puntos > mejor) {
      mejor = puntos
    }
    i += 1
  }
  mejor
}

//RONDA 1
println("--- RONDA 1 ---")
var i1 = 0
while (i1 < jugadores.length) {
  val jugador = jugadores(i1)
  val puntos = ronda1(i1)
  println(jugador + " tiene " + puntos + " puntos. ¿Se pasa? " + bust(puntos))
  i1 += 1
}
println("Mejor puntuación válida Ronda 1: " + mejorPuntuacion(ronda1))

// RONDA 2
println("\n--- RONDA 2 ---")
var i2 = 0
while (i2 < jugadores.length) {
  val jugador = jugadores(i2)
  val puntos = ronda2(i2)
  println(jugador + " tiene " + puntos + " puntos. ¿Se pasa? " + bust(puntos))
  i2 += 1
}
println("Mejor puntuación válida Ronda 2: " + mejorPuntuacion(ronda2))
````

**Informe final del ejercicio (Ejercicio 15):**

* **Estructuras de datos:** Usamos una `List` inmutable para los nombres y un `Array[Int]` para las puntuaciones de cada ronda sin modificar los originales.
